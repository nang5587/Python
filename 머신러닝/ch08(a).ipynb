```python
from sklearn.datasets import load_files

review_train = load_files("data/aclImdb/train") # iris 데이터처럼 나에게 줄것임
text_train, y_train = review_train.data[:1000], review_train.target[:1000]
text_train = [doc.replace(b"<br />", b" ")for doc in text_train] # 정규식 : 챗에게 물어보자
# 반복문보다 List comprehensive가 빠름
```

```python
review_test = load_files("data/aclImdb/test")
text_test, y_test = review_test.data, review_test.target
text_test = [doc.replace(b"<br />", b" ")for doc in text_test]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

bards_words = [
    "The fool doth think he is wise, ",
    "but the wise man knows himself to be a fool"
]

vect = CountVectorizer()
vect.fit(bards_words)
bag_of_words = vect.transform(bards_words) # 희소행렬 
print(bag_of_words.toarray()) # 길면 길수록 행렬의 차수가 무한정 늘어남
```

```python
vec = CountVectorizer(max_features=10000, max_df=0.5) # 텍스트 데이터는 균질하게 데이터가 있는 게 좋음(다른 것과 다르게), 하나에 몰빵되면 오판하는 경우가 생김
# max_df=0.5 : 어떤 단어가 빈도 50%를 넘으면 빼겠다는 뜻
X = vec.fit_transform(text_train)
```

```python
from sklearn.decomposition import LatentDirichletAllocation

lda = LatentDirichletAllocation(n_components=10, learning_method="batch", max_iter=25, random_state=42) # n_components=10 : 중요할 거 같은거 10개 (순서 없음) -> 나오면 내가 판단해야 함
document_topics = lda.fit_transform(X)
```

```python
import numpy as np

sorting = np.argsort(lda.components_, axis=1)[:, ::-1]
feat_names = np.array(vec.get_feature_names_out())

for topic_idx, topic in enumerate(lda.components_):
    top_feats = feat_names[sorting[topic_idx, :10]]
    print(f"토픽: {topic_idx} 주요 단어: {top_feats}")
```

```python
lda00 = LatentDirichletAllocation(n_components=100, learning_method="batch", max_iter=25, random_state=42) # n_components=10 : 중요할 거 같은거 10개 (순서 없음) -> 나오면 내가 판단해야 함
document_topics00 = lda00.fit_transform(X)
# batch : sklearn으로 읽을 수 있으면 일반적으로 batch 쓴다고 보면 됨
```

```python
sorting = np.argsort(lda00.components_, axis=1)[:, ::-1]
feat_names = np.array(vec.get_feature_names_out())
topic_names = [
    f"{i:>2} " + " ".join(words)
    for i, words in enumerate(feat_names[sorting[:, :2]])
]
topic_names
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

scores = cross_val_score(LogisticRegression(max_iter=1000), X, y_train, n_jobs=-1)
print(scores)
```

# AND 연산을 활용한 머신러닝 연습

## 기울기와 절편
y = aX + b
- a: 기울기(coef_, weight)
- b: 절편(intercept_, bias)

```python
import numpy as np
```

```python
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])
```

```python
weight = np.random.randn(2) # 초기값 설정 일단 랜덤으로
bias = np.random.randn()
learning_rate = 0.1
epochs = 100
```

```python
# 1차 완료
for epoch in range(epochs):
    z = np.dot(X, weight) + bias
    predictions = np.where(z>=0, 1, 0) # 0보다 크면 1나오고 아니면 0나옴
    errors = y - predictions
    weight += learning_rate * np.dot(X.T, errors)
    bias += learning_rate * np.sum(errors)
    accuracy = np.mean(predictions == y)
    # if epoch % 10 == 0:
    #     print(f'{epoch}, {accuracy:.4f}')
```

```python
print(weight, bias)
```

```python
for i in range(len(X)):
    z= np.dot(X[i], weight) + bias
    prediction = 1 if z>=0 else 0
    print(f'{X[i]}, {prediction}, {y[i]}')
```

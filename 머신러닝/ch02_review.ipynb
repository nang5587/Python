```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier # 분류 문제니까 얘 들고와야 함
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# 1st
df = pd.read_csv('data/penguins.csv')
df.head()
```

```python
df['species'].value_counts() # 답안지 구성 확인
```

```python
# 2nd
# df = pd.get_dummies(df, columns=['species', 'sex', 'island'], drop_first=True) # 강제로 타입 변환시킴, bool로 0, 1 int로 변환되니까
df.head()
```

```python
# 3rd
df = df.fillna(0) # 원래는 이러면 안 되고 중위값이나 평균값 넣어야 함
df.info()
```

```python
X = df.drop(['species', 'sex', 'island', 'rowid'], axis=1)
y = df["species"]
print(X.shape, y.shape)
```

```python
#---------------knn으로 학습
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
knn = KNeighborsClassifier(n_neighbors=5) # n_neighbors 바꿔가면서 결과 확인해서 좋은 거 찾아보기
knn.fit(X_train, y_train)
```

```python
lr = LogisticRegression(max_iter=3000)
lr.fit(X_train, y_train)
```

```python
gnb = GaussianNB()
gnb.fit(X_train, y_train)
```

```python
knn_y_pred = knn.predict(X_test)
lr_y_pred = lr.predict(X_test)
gnb_y_pred = gnb.predict(X_test)
```

```python
print(accuracy_score(knn_y_pred, y_test))
print(accuracy_score(lr_y_pred, y_test)) # 선형이 더 좋을 수도
print(accuracy_score(gnb_y_pred, y_test))
# 분류는 맞는 걸 맞다고 한 것이랑 틀린 걸 틀리다고 하는 게 중요
```

```python
class Point :
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def get_x(self):
        return self.x
    def get_y(self):
        return self.y
    def set_x(self, x):
        self.x =x
    def set_y(self,y):
        self.y =y
    
    
        return self.y
```

```python
p = Point(1, 2)
type(p)
p.set_x(5)
p.set_y(10)
p.get_x()
```

```python
class Rectangle(Point):
    def __init__(self, x, y):
        super().__init__(x, y)

r = Rectangle(1,2)
type(r)
r.get_x()
```

```python
class Person:
    def __init__(self, name):
        self.name = name

    def display(self):
        msg = f'저는 {self.name}입니다.'
        print(msg)

p = Person('강나현')
p.display()
```

```python
class Student(Person):
    def __init__(self, name, student_id):
        super().__init__(name)
        self.student_id = student_id
    
    def display(self):
        super().display()
        msg =  f'학번은 {self.get_student_id()}입니다.'
        print(msg)

    def get_student_id(self):
        return self.student_id

s = Student('강나현', 2145587)
s.display()
```

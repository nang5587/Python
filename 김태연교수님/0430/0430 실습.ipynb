실습 문제 1: ReLU 함수 (Rectified Linear Unit)
설명: ReLU 함수 ( f(x) = \max(0, x) )는 딥러닝에서 가장 널리 사용되는 비선형 활성화 함수입니다. 입력값이 0보다 작으면 0을, 0 이상이면 입력값 그대로를 출력합니다. 이 함수의 동작을 시각화해보세요.

{\displaystyle \operatorname {ReLU} (x)=x^{+}=\max(0,x)={\frac {x+|x|}{2}}={\begin{cases}x&{\text{if }}x>0,\\0&x\leq 0\end{cases}}}

요구사항:
( x )의 범위를 -5부터 5까지 설정하고, NumPy를 사용해 ReLU 함수 값을 계산하세요.
Matplotlib으로 그래프를 그리고, x축 레이블을 "x", y축 레이블을 "f(x)", 제목을 "ReLU Function"으로 설정하세요.
그래프에서 ( x = 0 )에서의 꺾이는 지점을 관찰하세요.

```python
import numpy as np
import matplotlib.pyplot as plt

def relu(x):
    # if x < 0 : return 0

    return (x + np.abs(x))/2

x = np.linspace(-5, 5, 100)
plt.plot(x, relu(x))
plt.show()
```

1. 정수 N개가 저장된 배열 numbers가 주어집니다. 또 다른 정수 M개가 저장된 배열 targets가 주어집니다. targets 배열의 각 원소가 numbers 배열에 몇 번 등장하는지 계산하여 출력하는 프로그램을 작성하십시오.

```python
import collections
from typing import List, Tuple

def count_numbers(N: int, numbers: List[int], M: int, targets: List[int]) -> List[int]:
    lst = []
    for i in range(M):
        lst.append(numbers.count(targets[i]))
    return lst

input_data = [
    (10, [6, 3, 2, 10, 10, 10, -10, -10, 7, 3], 8, [10, 9, -5, 2, 3, 4, 5, -10]), # 예제 입력 1
    (5, [1, 1, 1, 2, 3], 4, [1, 2, 3, 4]),
    (3, [-1, 0, 1], 3, [0, 1, -1]),
    (1, [5], 2, [5, 0]),
    (4, [0, 0, 0, 0], 2, [0, 1]),
    (6, [1, 2, 3, 4, 5, 6], 6, [6, 5, 4, 3, 2, 1])
]

output_data = [
    [3, 0, 0, 1, 2, 0, 0, 2],
    [3, 1, 1, 0],
    [1, 1, 1],
    [1, 0],
    [4, 0],
    [1, 1, 1, 1, 1, 1]
]

# 테스트 실행
for i, ((N, numbers, M, targets), expected_output) in enumerate(zip(input_data, output_data)):
    result = count_numbers(N, numbers, M, targets)
    assert result == expected_output, \
        f"Test Case {i+1} Failed:\nInput: N={N}, numbers={numbers}, M={M}, targets={targets}\nExpected: {expected_output}\nGot: {result}"
```

2. 쇼핑카트를 나타내는 ShoppingCart 클래스를 작성하세요. 이 클래스는 아이템을 추가, 제거하고 전체 가격을 계산하는 기능을 가져야 합니다. 아이템은 딕셔너리로 저장하며, 딕셔너리의 키는 아이템 이름, 값은 가격과 수량을 포함하는 튜플입니다.

Methods: init(): 아이템 이름을 키로 하고 (가격, 수량) 튜플을 값으로 가지는 딕셔너리를 초기화합니다. ShoppingCart 인스턴스가 생성될 때, 카트는 비어 있는 상태로 시작합니다.

add_item(item_name: str, price: float, quantity: int): 이 메서드는 지정된 수량만큼 아이템을 카트에 추가합니다. 만약 이미 카트에 있는 아이템이라면, 해당 아이템의 수량을 증가시킵니다.

remove_item(item_name: str, quantity: int): 카트에서 지정된 수량만큼 아이템을 제거합니다. 제거할 수량이 현재 남아있는 수량보다 크거나 같으면, 해당 아이템을 카트에서 완전히 삭제합니다.

get_total_price(): 카트에 있는 모든 아이템을 순회하며 각 아이템의 가격과 수량을 곱해 총 비용을 계산합니다.

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, item_name: str, price: float, quantity: int):
        if item_name in self.items :
            ori_price, ori_quantiry = self.items[item_name]
            self.items[item_name] = (ori_price, ori_quantiry + quantity)
        else:     
            self.items[item_name] = (price, quantity)

    def remove_item(self, item_name: str, quantity: int):
        if item_name in self.items :
            ori_price, ori_quantiry = self.items[item_name]
            # if ori_quantiry - quantity > 0 :
            self.items[item_name] = (ori_price, ori_quantiry - quantity)

    def get_total_price(self) -> float:
        total = 0
        for ori_price, ori_quantiry in self.items.items():
            pr , qu = ori_quantiry
            price = int(pr)
            total = total + price
            
        return total/len(self.items)
```

```python
input_data_cart = [
    [("add", "apple", 1.5, 10), ("add", "banana", 0.5, 5), ("add", "apple", 1.5, 5)],  # Adding items, updating apple
    [("remove", "banana", None, 2), ("remove", "apple", None, 12), ("add", "orange", 2.0, 2)],  # Removing and adding more
]

expected_cart_states = [
    {"apple": (1.5, 15), "banana": (0.5, 5)},
    {"orange": (2.0, 2)}
]

expected_total_prices = [25.0, 4.0]

for data_set, expected_cart, expected_total in zip(input_data_cart, expected_cart_states, expected_total_prices):
    cart = ShoppingCart()

    for action, item_name, price, quantity in data_set:
        if action == "add":
            cart.add_item(item_name, price, quantity)
        elif action == "remove":
            cart.remove_item(item_name, quantity)

    assert cart.items == expected_cart

    total_price = cart.get_total_price()
    assert total_price == expected_total
```

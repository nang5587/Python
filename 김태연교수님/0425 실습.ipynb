실습 문제 3: 삼각 함수 (Sine)

함수 f(x) = sin(x) 를 구현하고 그래프를 그려보세요.

요구사항:
0부터 2π (약 6.28) 까지의 x 값에 대해 y 값을 계산하는 함수를 순수 Python으로 작성하세요. (힌트: math.sin 사용, 라디안 단위 사용)
동일한 기능을 하는 함수를 NumPy를 사용하여 작성하세요. (힌트: np.sin, np.pi, np.linspace)
두 함수로부터 얻은 결과를 Matplotlib으로 시각화하세요.

```python
import numpy as np
import matplotlib.pyplot as plt

def sin(x):
    return np.sin(x)

x = np.linspace(0, 2*(np.pi),50)
plt.plot(x, sin(x))
plt.show()
```

실습 문제 4: 지수 함수 (Exponential)

자연 상수 e의 거듭제곱 함수 f(x) = e^x 를 구현하고 그래프를 그려보세요. 이 함수는 AI에서 확률 계산 등에 사용될 수 있습니다.

요구사항:
2부터 2까지의 x 값에 대해 y 값을 계산하는 함수를 순수 Python으로 작성하세요. (힌트: math.exp)
동일한 기능을 하는 함수를 NumPy를 사용하여 작성하세요. (힌트: np.exp)
두 함수로부터 얻은 결과를 Matplotlib으로 시각화하세요.

```python
import numpy as np
import matplotlib.pyplot as plt

def expo(x):
    return np.exp(x)

x = np.linspace(-2, 2)
plt.plot(x, expo(x))
plt.show()
```
